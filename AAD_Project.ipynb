#### 1.2.2 Valori aberante (Outliers)

**Ce trebuie sa faceti:**
1. Identificati outliers folosind IQR sau Z-score
2. Vizualizati distributia (boxplot)
3. Decideti daca eliminati/pastrati (argumentati!)

```python
def get_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 2.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    return outliers, lower, upper
```

```python
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 2.5 * IQR


    df_clean = df.copy()
    df_clean[column] = np.where((df[column] < lower), lower, df[column])
    df_clean[column] = np.where((df[column] > upper), upper, df[column])

    return df_clean
```

For the majority of the datasets we are going to replace the outliers with the closest valid values, with the only exception being "Meat Consumption" because of the way that diferent religions/culture promote or vorbid eating diferent types of meat. At the same time we are going to use as the upper limit the value of 2.5 * IQR, as we believe that there is a reasonable posibility for countries to be way poorer than others for example.

```python
cleaned_datasets = {}

for name, dataset in datasets.items():
    print(f"\nProcessing dataset: {name}")
    display(dataset.head(10))

    num_cols = dataset.select_dtypes(include='number').columns

    dataset = dataset.reset_index(drop=True)
    df_clean = dataset.copy()

    display(dataset.head(10))
    display(num_cols)

    for col in num_cols:
        if dataset[col].nunique() < 5 or col == 'Year':
            continue

        outliers, lower, upper = get_outliers_iqr(dataset, col)

        plt.figure(figsize=(12, 4))
        plt.plot(dataset.index, dataset[col], 'b.', alpha=0.4, label="Data")

        plt.scatter(outliers.index,
                    outliers[col],
                    c='red', s=20, label="Outliers")

        plt.axhline(lower, color='orange', linestyle='--')
        plt.axhline(upper, color='orange', linestyle='--')

        plt.title(f"{name} - {col} outliers")
        plt.legend()
        plt.show()

        if name != 'Meat consumption':
            df_clean = remove_outliers_iqr(df_clean, col)

    cleaned_datasets[name] = df_clean

for name, dataset in cleaned_datasets.items():
    print(f"\nDataset: {name}")
    display(dataset.head())
    display(dataset.shape)
```

```python
#def statistics(df):
  # TODO: Statistici descriptive
  # df.describe()

  # TODO: Distributia variabilelor categorice
  # df['coloana_categorica'].value_counts()

  # TODO: Matrice de corelatie
  # corr_matrix = df.select_dtypes(include=[np.number]).corr()
  # sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
  # pass

from itertools import combinations
from scipy.stats import chi2_contingency

def analyze_dataframe(df, top_n=5):
    print("=== DESCRIPTIVE STATISTICS ===")
    print(df.describe(include='all'))

    print("\n=== CATEGORICAL DISTRIBUTIONS ===")
    cat_cols = df.select_dtypes(include=['object', 'category']).columns
    for col in cat_cols:
        print(f"\nColumn: {col}")
        print(df[col].value_counts())

    print("\n=== NUMERICAL CORRELATION MATRIX ===")
    num_df = df.select_dtypes(include=[np.number])
    if not num_df.empty:
        corr_matrix = num_df.corr()
        print(corr_matrix)

        plt.figure(figsize=(10, 8))
        sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
        plt.title("Correlation Heatmap")
        plt.show()

        # Extract top correlations
        corr_pairs = (
            corr_matrix.abs()
            .unstack()
            .sort_values(ascending=False)
        )
        corr_pairs = corr_pairs[corr_pairs < 1.0]  # remove self-correlation

        print(f"\nTop {top_n} numerical correlations:")
        print(corr_pairs.head(top_n))
    else:
        print("No numerical columns found.")

    print("\n=== CHI-SQUARE TEST (CATEGORICAL VARIABLES) ===")
    chi_results = []

    for col1, col2 in combinations(cat_cols, 2):
        contingency_table = pd.crosstab(df[col1], df[col2])

        if contingency_table.shape[0] > 1 and contingency_table.shape[1] > 1:
            chi2, p, dof, expected = chi2_contingency(contingency_table)
            chi_results.append((col1, col2, chi2, p))

    # Sort by chi2 statistic (higher = stronger relationship)
    chi_results_sorted = sorted(chi_results, key=lambda x: x[2], reverse=True)

    print(f"\nTop {top_n} categorical correlations (Chi-square):")
    for col1, col2, chi2, p in chi_results_sorted[:top_n]:
        print(f"{col1} vs {col2} -> chi2={chi2:.2f}, p-value={p:.4f}")
```

```python
def smart_correlation_analysis(df, threshold=0.6):
    num_df = df.select_dtypes(include=[np.number])
    cols = num_df.columns

    # --- AUTO GROUPING BASED ON COLUMN NAMES ---
    groups = {
        "Health": [c for c in cols if any(k in c.lower() for k in ["mortality", "death"])],
        "Economic": [c for c in cols if any(k in c.lower() for k in ["income", "gni", "gdp"])],
        "Diet": [c for c in cols if any(k in c.lower() for k in ["diet", "consumption", "calories"])],
        "Food Security": [c for c in cols if any(k in c.lower() for k in ["cost", "afford", "poverty"])]
    }

    # Remove empty groups
    groups = {k: v for k, v in groups.items() if len(v) > 1}

    corr = num_df.corr()

    # --- 1. HEATMAP PER GROUP ---
    for group_name, group_cols in groups.items():
        sub_corr = corr.loc[group_cols, group_cols]

        plt.figure(figsize=(8, 6))
        sns.heatmap(sub_corr, cmap='coolwarm', annot=False)
        plt.title(f"{group_name} Correlations")
        plt.xticks(rotation=45, ha='right')
        plt.yticks(rotation=0)
        plt.show()

    # --- 2. CROSS-GROUP CORRELATIONS ---
    print("\n=== CROSS-GROUP STRONG CORRELATIONS ===")
    for g1, cols1 in groups.items():
        for g2, cols2 in groups.items():
            if g1 >= g2:
                continue

            sub_corr = corr.loc[cols1, cols2]

            strong = sub_corr[
                sub_corr.abs() >= threshold
            ].stack().sort_values(ascending=False)

            if not strong.empty:
                print(f"\n{g1} vs {g2}:")
                print(strong.head(5))

    # --- 3. TOP GLOBAL CORRELATIONS ---
    print("\n=== TOP GLOBAL CORRELATIONS ===")
    corr_pairs = corr.unstack()
    corr_pairs = corr_pairs[
        (corr_pairs.abs() >= threshold) & (corr_pairs.abs() < 1)
    ].sort_values(ascending=False)

    print(corr_pairs.drop_duplicates().head(10))

smart_correlation_analysis(df_merged)
```

```python
column_list = ['Diet Composition Sugar', 'Diet Composition Oils And Fats',
       'Diet Composition Meat', 'Diet Composition Dairy And Eggs',
       'Diet Composition Fruit And Vegetables', 'Diet Composition Cereals And Grains']

x = df_merged['Income per Capita'].values.reshape(-1, 1)
for column in column_list:
    y = df_merged[column].values

    poly = PolynomialFeatures(degree=2)
    x_poly = poly.fit_transform(x)

    poly_reg = LinearRegression()
    poly_reg.fit(x_poly, y)

    x_range = np.linspace(np.min(x), np.max(x), 100).reshape(-1, 1)
    x_range_poly = poly.transform(x_range)
    y_pred = poly_reg.predict(x_range_poly)

    fig = px.scatter(df_merged, x='Income per Capita', y=column,
                 title='Corelation between Incompe per Capita and ' + column,
                 labels={'Incompe per Capita': 'Income per Capita ($)', column: column },
                 opacity=0.5)

    coefficients = poly_reg.coef_
    intercept = poly_reg.intercept_

    a = coefficients[2]
    b = coefficients[1]
    c = intercept

    fig.add_trace(go.Scatter(x=x_range.flatten(), y=y_pred,
                         mode='lines',
                         name=f'Fit: y = {a:.4f}x² + {b:.4f}x + {c:.2f}',
                         line=dict(color='red', width=2)))

    fig.show()
```

```python
# Calculati pentru Model 1:
print("=== Model 1: Linear Regression ===")

model1 = LinearRegression()
model1.fit(X_train_scaled, y_train)

y_pred_1 = model1.predict(X_test_scaled)
r2_1 = r2_score(y_test, y_pred_1)
rmse_1 = np.sqrt(mean_squared_error(y_test, y_pred_1))
mae_1 = mean_absolute_error(y_test, y_pred_1)

print(f"R2: {r2_1:.4f}")
print(f"RMSE: {rmse_1:.4f}")
print(f"MAE: {mae_1:.4f}")
```

```python
# Calculati pentru Model 2:
print("=== Model 2: Random Forest ===")

r2s = []
rmses = []
maes = []
depths = []

for i in range(6, 17, 2):
    model2 = RandomForestRegressor(max_depth=i, random_state=42)
    model2.fit(X_train, y_train)
    depths.append(i)

    y_pred_2 = model2.predict(X_test)
    r2_2 = r2_score(y_test, y_pred_2)
    rmse_2 = np.sqrt(mean_squared_error(y_test, y_pred_2))
    mae_2 = mean_absolute_error(y_test, y_pred_2)
    r2s.append(r2_2)
    rmses.append(rmse_2)
    maes.append(mae_2)

    print(f"This is for depth: {i}")
    print(f"R2: {r2_2:.4f}")
    print(f"RMSE: {rmse_2:.4f}")
    print(f"MAE: {mae_2:.4f}")
    print("\n")

max = np.argmax(r2s)
r2_2 = r2s[max]
rmse_2 = rmses[max]
mae_2 = maes[max]
depth = depths[max]
print(depth)
```

```python
print("=== Model 3: Gradient Boosting ===")

r2s = []
rmses = []
maes = []
depths = []


for estimators in range(50, 151, 50):
    for rate in [0.05, 0.1, 0.15]:
        for depth in range(3, 10, 3):
            model3 = GradientBoostingRegressor(n_estimators=estimators, learning_rate=rate, max_depth=depth, random_state=42)
            model3.fit(X_train, y_train)
            depths.append((estimators, rate, depth))

            y_pred_3 = model3.predict(X_test)
            r2_3 = r2_score(y_test, y_pred_3)
            rmse_3 = np.sqrt(mean_squared_error(y_test, y_pred_3))
            mae_3 = mean_absolute_error(y_test, y_pred_3)
            r2s.append(r2_3)
            rmses.append(rmse_3)
            maes.append(mae_3)

            print(f"This is for parameters: {(estimators, rate, depth)}")
            print(f"R2: {r2_3:.4f}")
            print(f"RMSE: {rmse_3:.4f}")
            print(f"MAE: {mae_3:.4f}")
            print("\n")

max = np.argmax(r2s)
r2_3 = r2s[max]
rmse_3 = rmses[max]
mae_3 = maes[max]
parametersGradient = depths[max]
print(parametersGradient)
```

```python
import xgboost as xgb
import re

print("=== Model 4: XGBoost ===")
xgb_model = xgb_model = xgb.XGBRegressor(
    objective='reg:squarederror',
    random_state=42,
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=1.0,
)

regex = re.compile(r"\[|\]|<", re.IGNORECASE)
X_train.columns = [regex.sub("_", col) if any(x in str(col) for x in set(('[', ']', '<'))) else col for col in X_train.columns.values]
X_test.columns = [regex.sub("_", col) if any(x in str(col) for x in set(('[', ']', '<'))) else col for col in X_test.columns.values]

xgb_model.fit(X_train, y_train)

y_pred_4 = xgb_model.predict(X_test)

r2_4 = r2_score(y_test, y_pred_4)
rmse_4 = np.sqrt(mean_squared_error(y_test, y_pred_4))
mae_4 = mean_absolute_error(y_test, y_pred_4)

print(f"R2: {r2_4:.4f}")
print(f"RMSE: {rmse_4:.4f}")
print(f"MAE: {mae_4:.4f}")
```
